For a function in quadratic form $q(w) = \frac12x^TQx+b^Tx+c$, $q(w)$ is convex iff $Q$ is PSD, and is strictly convex when $Q$ is PD. expanding $f(w)$:
$$
\begin{align}
f(w) &= \frac12||(e\:X)w-y||^2 \\
&= \frac12((e\:X)w-y)^T((e\:X)w-y) \\
&= \frac12(w^T(e\:X)^T-y^T)((e\:X)w-y) \\
&= \frac12(w^T(e\:X)^T(e\:X)w) -y^T(e\:X)w +\frac12y^Ty.
\end{align}
$$
As $w^T(e\:X)^T(e\:X)w = ((e\:X)w)^T((e\:X)w) = ||(e\:X)w||^2 \geq 0$, $\forall w \in \mathbb{R}^{p+1}$, $Q=(e\:X)^T(e\:X)$ is PSD and thus $f(w)$ is conve\:X.

To ensure $Q$ is PD, we would need $((e\:X)w)^T((e\:X)w) = ||(e\:X)w||^2 > 0$ for all $w \neq 0$. This requirement is the definition of linear independence, so we must have $e\:X$ be linearly independent. This occurs when $X$ is linearly independent and $e$ is not a linear combination of the columns of $X$. 

It is almost certain that the wine dataset satisfies this condition. As there is a positive amount of variance with the values for each column, it isn't possible that one entire column is constant (multiple of $e$). Moreover, it is extremely unlikely that there is a linear combination of columns that results in a constant column, since each value's range is continuous, and there are over 4000 observations. 

## 2.

Let $t_k$ be step size at $k$th iteration. Then

$$
\begin{align}
t_k &= \text{arg min}_t(f(w^{(k)}-t\nabla_k) \\
&= \text{arg min}_t(\frac12||(e\:X)w^{(k)}-y-t(e\:X)\nabla_k||^2)
\end{align}
$$

To find the minimum, we take the derivative of $f$ w.r.t. to $t$ and find the zeros:
$$
\begin{align}
0 &= \frac{d}{dt}(\frac12||(e\:X)w^{(k)}-y-t(e\:X)\nabla_k||^2) \\
&= ((e\:X)w^{(k)}-y-t(e\:X)\nabla_k)^T((e\:X)\nabla_k) \\
\end{align}
$$
Solving,
$$
\begin{align}
t((e\:X)\nabla_k)^T(e\:X)\nabla_k &= ((e\:X)w^{(k)}-y)^T(e\:X)\nabla_k \\
t\nabla_k^T(e\:X)^T(e\:X)\nabla_k &= ((e\:X)^T((e\:X)w^{(k)}-y))^T\nabla_k \\
t &= \frac{(\nabla_k)^T\nabla_k}{\nabla_k^T(e\:X)^T(e\:X)\nabla_k}
\end{align}
$$
To arrive at the final line above, note that $\nabla_k = (e\:X)^T(e\:X)w^{(k)}-(e\:X)^Ty$, which we substitute into the numerator.

```python
import numpy as np
import time

data = np.loadtxt("winequality-white.csv", delimiter=';', skiprows=1)
X = data[:, :-1]
y = data[:, -1:].reshape(-1, 1)

X_train, X_test = X[:4298], X[4298:]
y_train, y_test = y[:4298], y[4298:]
```

```python
n, p = X_train.shape
e = np.ones((n, 1))
eX = np.hstack([e, X_train])

w = np.zeros((p + 1, 1))
tolerance = 1e-6
iter = 5000
```

```python
#printing only first 10 iterations to avoid clutter in final submission

print("\n    iter     w[0]         w[1]       ||grad||      step-len")
print("-------------------------------------------------------------")
start = time.time()
for i in range(iter):
    r = eX @ w - y_train
    fun = 0.5 * float(r.T @ r)
    grad = eX.T @ r
    d = -grad
    normd = np.linalg.norm(d)
    if(i<10):
        print(f"{i:8d} {float(w[0]):11.6f} {float(w[1]):11.6f} {normd:12.3e}", end="")
    if(np.linalg.norm(grad) < tolerance):
        break
    t = float((grad.T @ grad) / (grad.T @ eX.T @ eX @ grad))
    if(i<10):
        print(f"   {t:11.6f}")
    w = w + t*d
end = time.time()
w_star = w
runtime = end-start
```

```python
w_star
```

```python
r_final = eX @ w_star - y_train
fun_final = 0.5 * float(r_final.T @ r_final)
grad_final = eX.T @ r_final
norm_final = np.linalg.norm(grad_final)
```

```python
Q = eX.T @ eX
eigvals = np.linalg.eigvalsh(Q)
kappa = eigvals.max() / eigvals.min()
rho = ((kappa-1)/(kappa+1))**2
```

```python
w_star, rho, f"{runtime:.4f}", f"{fun_final:.4f}", f"{norm_final:.4f}"
```

```python
y_pred_train = eX @ w_star

eX_test = np.hstack([np.ones((X_test.shape[0], 1)), X_test])
y_pred_test = eX_test @ w_star

mae_train = np.mean(np.abs(y_train - y_pred_train))
mae_test = np.mean(np.abs(y_test - y_pred_test))
mae_train, mae_test
```
